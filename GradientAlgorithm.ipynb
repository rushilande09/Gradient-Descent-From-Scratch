```python
import numpy as  np
import matplotlib.pyplot as plt
```


## 1. Dataset Initialization

We use a simple dataset containing seven observations.

The independent variable (X) and dependent variable (Y)
have a linear relationship.

The dataset follows the relationship:

y = 2x + 1

We initialize the model parameters theta0 and theta1
with zero values.


## 2. Hypothesis Function

The hypothesis function estimates the output
using the current model parameters.

h(x) = theta0 + theta1 * x

Here, theta0 is the intercept and theta1 is
the slope of the regression line.

```python
# Creating the hypothesis.
x = np.array([1,2,3,4,5,6,7])
y = np.array([3,5,7,9,11,13,15])

theta0 = 0
theta1 = 0
alpha = 0.01
m = len(x)

def hypothesis(theta0, theta1, x):
  return theta0 + theta1*x

print(hypothesis(0, 0, x))
print(hypothesis(1, 2, x))
```

```python
#Writing the cost function

actual = np.array([3, 5, 7])
predicted = np.array([2, 6, 8])

def cost_function(actual, predicted):
  error = np.subtract(actual, predicted)
  m = len(actual)

  return 1/2*1/m * np.sum(error**2)

cost_function(actual, predicted)
```


## 3. Partial Derivatives

Gradient Descent requires calculating the partial
derivatives of the cost function with respect to
theta0 and theta1.

Gradient with respect to theta0:

dJ/dtheta0 = (1/m) * sum(h(x) - y)

Gradient with respect to theta1:

dJ/dtheta1 = (1/m) * sum((h(x) - y) * x)

These gradients indicate the direction in which
the model parameters should be updated.

## 4. Cost Function

We use Mean Squared Error with a factor of 1/2
to measure the prediction error.

J(theta0, theta1) = (1/2m) * sum((h(x) - y)^2)

A lower cost indicates that the model predictions
are closer to the actual values.

The objective of Gradient Descent is to minimize
this cost function.

## 5. Gradient Descent Implementation

Gradient Descent is an optimization algorithm that
updates model parameters iteratively to minimize
the cost function.

Parameter update rules:

theta0 = theta0 - alpha * dJ/dtheta0

theta1 = theta1 - alpha * dJ/dtheta1

Where:
- alpha is the learning rate.
- The gradients determine the update direction.
- Iterations determine how many updates are performed.

We use a learning rate of 0.01 and 1000 iterations.

```python
#Implementing the Gradient Algorithm

theta0 = 0
theta1 = 0
alpha = 0.01
iterations = 1000
x_data = np.array([1, 2, 3, 4, 5, 6, 7])
y = np.array([3, 5, 7, 9, 11, 13, 15])
m = len(x_data)
cost_history= []

def hypothesis(theta0,theta1,x_data):
        h_x = theta0 + theta1 * x_data
        return h_x

def predicted(theta0,theta1,x_data):
      return hypothesis(theta0,theta1,x_data)

def gradient_descent_theta0(h_x,y):
        errors = np.subtract(h_x,y)
        gradient0 = 1/m*(np.sum(errors))
        return gradient0

    
def gradient_descent_theta1(h_x,y,x_data):
        errors = np.subtract(h_x,y)
        gradient1 = 1/m*(np.sum(errors*x_data))
        return gradient1

def cost_function(y,predicted):
    error = np.subtract(y, predicted)
     

    return 1/2*1/m * np.sum(error**2)

initial_cost = cost_function(y, hypothesis(theta0, theta1, x_data))

for i in range(iterations):
    h = hypothesis(theta0,theta1,x_data)
    ans0 = gradient_descent_theta0(h,y)
    ans1 = gradient_descent_theta1(h,y,x_data)
    theta0_up = theta0 - alpha * ans0
    theta1_up = theta1 - alpha * ans1

    theta0 = theta0_up
    theta1 = theta1_up

    
    cost_history.append(cost_function(y,predicted(theta0,theta1,x_data)))


print(theta0,theta1)



y_predicted = hypothesis(theta0,theta1,x_data)
final_cost = cost_function(y,predicted(theta0,theta1,x_data))



print("Predicted:",y_predicted)
print("Initial Cost:", initial_cost)
print("Final Cost:", cost_history[-1])





    
```

```python
plt.plot(cost_history)
plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.title("Gradient Descent Convergence")
plt.show()
```

```python
plt.scatter(x_data, y, color='blue',
            label='Actual Data', marker='o')

plt.plot(x_data, y_predicted, color='red',
         linestyle='--', label='Regression Line')

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression using Gradient Descent")
plt.legend()
plt.grid(True)
plt.show()
```
